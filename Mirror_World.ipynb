# Mirror World: Real-time Symmetry Drawing Tool
This notebook implements a drawing tool where your strokes are reflected across multiple axes of symmetry.

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, widgets
from IPython.display import display, clear_output

%matplotlib inline
```

```python
class MirrorCanvas:
    def __init__(self, symmetry_type='horizontal'):
        self.symmetry_type = symmetry_type
        self.points = []

    def add_point(self, x, y):
        new_points = [(x, y)]

        if self.symmetry_type == 'horizontal':
            new_points.append((x, -y))
        elif self.symmetry_type == 'vertical':
            new_points.append((-x, y))
        elif self.symmetry_type == 'quadrant':
            new_points.append((-x, y))
            new_points.append((x, -y))
            new_points.append((-x, -y))

        self.points.extend(new_points)

    def render(self):
        if not self.points:
            return
        pts = np.array(self.points)
        plt.figure(figsize=(6, 6))
        plt.scatter(pts[:, 0], pts[:, 1], s=10, c='blue')
        plt.xlim(-10, 10)
        plt.ylim(-10, 10)
        plt.axhline(0, color='black', lw=1)
        plt.axvline(0, color='black', lw=1)
        plt.title(f'Symmetry: {self.symmetry_type}')
        plt.show()
```

### Interaction Demo
Run the cell below to simulate drawing points. In a full implementation, we would use a JavaScript-based canvas for smoother mouse tracking, but this demonstrates the mathematical reflection.

```python
canvas = MirrorCanvas(symmetry_type='quadrant')

for i in range(1, 6):
    canvas.add_point(i, i*0.5)

canvas.render()
```

### Interactive Mirror World
Use the sliders below to move your 'pen' and watch the reflections appear in real-time.

```python
def interactive_mirror(symmetry, x, y):
    canvas = MirrorCanvas(symmetry_type=symmetry)
    steps = 10
    for i in range(steps + 1):
        px = (x * i) / steps
        py = (y * i) / steps
        canvas.add_point(px, py)

    canvas.render()

interact(
    interactive_mirror,
    symmetry=['horizontal', 'vertical', 'quadrant'],
    x=widgets.FloatSlider(min=-10, max=10, step=0.5, value=5),
    y=widgets.FloatSlider(min=-10, max=10, step=0.5, value=5)
);
```
